# CRM Leads — Exploratory Data Analysis (EDA)

**Raw dataset:** `C:\Users\user\Desktop\Group project\Data set\CRM Leads.csv`

This notebook only explores the raw data. It does not modify the original dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt



df = pd.read_csv(r"C:\Users\user\Desktop\Group project\Data set\CRM Leads.csv")

print("Data loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## 1. Preview Raw Data

```python
display(df.head())
display(df.tail())
```

## 2. Dataset Structure

```python
print("Shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()
```

## 3. Statistical Summary

```python
display(df.describe(include='all').T)
```

## 4. Missing-Value Analysis

```python
missing = pd.DataFrame({
    'Missing_Count': df.isna().sum(),
    'Missing_Percentage': df.isna().mean() * 100
}).sort_values('Missing_Percentage', ascending=False)

display(missing[missing['Missing_Count'] > 0])

plt.figure(figsize=(12, 6))
m = missing[missing['Missing_Count'] > 0].head(20).sort_values('Missing_Percentage')
plt.barh(m.index, m['Missing_Percentage'])
plt.xlabel('Missing Percentage (%)')
plt.title('Top 20 Columns by Missing Percentage')
plt.tight_layout()
plt.show()
```

## 5. Duplicate Analysis

```python
print("Duplicate rows:", df.duplicated().sum())

if 'Lead ID' in df.columns:
    print("Duplicate Lead IDs:", df['Lead ID'].duplicated().sum())
    display(df[df['Lead ID'].duplicated(keep=False)].head(20))
```

## 6. Unique Values / Categorical Analysis

```python
categorical_candidates = [
    'Lead Stage', 'Lead Quality', 'Lead Source', 'Campaign Name',
    'Qualification', 'District', 'Preferred Location', 'Program',
    'Domain Interested', 'Coding Basics'
]

for col in categorical_candidates:
    if col in df.columns:
        print(f"\n--- {col} ---")
        print("Unique values:", df[col].nunique(dropna=True))
        display(df[col].value_counts(dropna=False).head(20))
```

## 7. Numerical Analysis

```python
numeric_candidates = [
    'Won Revenue', 'Call Duration',
    'Total_Outgoing_Calls', 'Total_Incoming_Calls'
]

for col in numeric_candidates:
    if col in df.columns:
        temp = pd.to_numeric(df[col], errors='coerce')
        print(f"\n--- {col} ---")
        print(temp.describe())
```

## 8. Date/Time Analysis

```python
date_candidates = [
    'Won Date', 'Lead Date', 'Expected Closing Date',
    'Lead Assigned At', 'First Touch At', 'Last Call Time',
    'Next Follow-up On', 'Created Time', 'Updated Time', 'Viewed Time'
]

for col in date_candidates:
    if col in df.columns:
        converted = pd.to_datetime(df[col], errors='coerce')
        print(f"{col}: valid={converted.notna().sum()}, missing/invalid={converted.isna().sum()}")
```

## 9. Lead Source Analysis

```python
if 'Lead Source' in df.columns:
    display(df['Lead Source'].value_counts(dropna=False).head(20))
    df['Lead Source'].value_counts(dropna=False).head(15).plot(kind='bar', figsize=(10,5))
    plt.title('Leads by Source')
    plt.ylabel('Number of Leads')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

## 10. Lead Stage Analysis

```python
if 'Lead Stage' in df.columns:
    display(df['Lead Stage'].value_counts(dropna=False))
    df['Lead Stage'].value_counts(dropna=False).plot(kind='bar', figsize=(9,5))
    plt.title('Leads by Stage')
    plt.ylabel('Number of Leads')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

## 11. Lead Quality Analysis

```python
if 'Lead Quality' in df.columns:
    display(df['Lead Quality'].value_counts(dropna=False))
    df['Lead Quality'].value_counts(dropna=False).plot(kind='bar', figsize=(8,5))
    plt.title('Leads by Quality')
    plt.ylabel('Number of Leads')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

## 12. Revenue Analysis

```python
if 'Won Revenue' in df.columns:
    revenue = pd.to_numeric(df['Won Revenue'], errors='coerce')
    print("Total Won Revenue:", revenue.sum())
    print("Average Won Revenue:", revenue.mean())
    print("Negative revenue records:", (revenue < 0).sum())
    display(df.loc[revenue < 0, ['Won Revenue']].head(20))
```
