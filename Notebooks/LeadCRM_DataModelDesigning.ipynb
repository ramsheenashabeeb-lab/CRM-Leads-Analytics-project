# Lead CRM – Relational Data Model Design

The supplied sample uses a student/advisor/reviewer star schema. For this dataset, the same modeling workflow is adapted to Lead CRM entities.

## Proposed tables

- `dim_user` — lead owner/user reference
- `dim_source` — lead source reference
- `dim_campaign` — campaign and advertising attributes
- `dim_contact` — contact and profile/education attributes
- `fact_lead` — one row per source lead
- `fact_lead_activity` — call/follow-up information
- `dim_date` — calendar attributes

### Relationships

`dim_user 1 → many fact_lead`  
`dim_source 1 → many fact_lead`  
`dim_campaign 1 → many fact_lead`  
`dim_contact 1 → many fact_lead`  
`fact_lead 1 → many fact_lead_activity`  
`dim_date 1 → many fact_lead`

## Modeling decisions

The source contains 10,000 rows and 52 columns.

The `Lead ID` column is completely empty, so the relational design creates a new integer `lead_id`.

The following columns are also completely empty in the supplied file and therefore are not stored:
`Lead Automation Email`, `Lead ID`, `Campaign Term`, `Campaign Content`, `Droped Date`, `Total_Outgoing_Calls`, and `Total_Incoming_Calls`.

Placeholder values such as `--Select--`, `--None--`, and `-Other-` are treated as missing during table preparation.
