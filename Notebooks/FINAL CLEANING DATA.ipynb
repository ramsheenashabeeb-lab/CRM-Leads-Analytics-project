# CRM Leads — Complete Data Cleaning Notebook

## Objective
Clean the CRM Leads dataset for analysis and Power BI.

### Rules used
1. Replace missing values wherever replacement is meaningful.
2. Do not delete rows because of missing values.
3. Convert date/time columns into the correct format.
4. Convert numeric columns into appropriate numeric data types.
5. Standardize text and categorical values.
6. Handle invalid negative revenue values.
7. Check duplicate rows without automatically deleting them.
8. Validate the dataset after cleaning.
9. Export the cleaned dataset for Power BI.

## Clean text/categorical columns

```python
df["Program"] = df["Program"].fillna("Unknown")
df["College"] = df["College"].fillna("Unknown")
df["Lead Source"] = df["Lead Source"].fillna("Unknown")
df["Lead Owner"] = df["Lead Owner"].fillna("Unknown")
df["Caller Name"] = df["Caller Name"].fillna("Unknown")
df["Call DID Status"] = df["Call DID Status"].fillna("Unknown")
df["Qualification"] = df["Qualification"].fillna("Unknown")
df["District"] = df["District"].fillna("Unknown")
df["Created By"] = df["Created By"].fillna("Unknown")
df["Lead Title"] = df["Lead Title"].fillna("Unknown")
df["Qualification"] = df["Qualification"].replace("--Select--", "Unknown")
df["Preferred Location"] = df["Preferred Location"].replace("-Other-", "Other")
df["Tomorrow closing"] = df["Tomorrow closing"].fillna("No")
```

## Replace placeholder values in Lost Reason

```python
df["Lost Reason"] = df["Lost Reason"].replace(
    ["--None--", "--None --", "None", ""],
    "Not Applicable"
)
# Replace actual missing values
df["Lost Reason"] = df["Lost Reason"].fillna("Not Applicable")
```

## Replace Invalid Values in lead quality 

```python
df["Lead Quality"] = df["Lead Quality"].fillna("Unknown")
df["Lead Quality"] = df["Lead Quality"].replace(
    ["", "--Select--", "nan"], "Unknown"
)
```

## Replace Invalid Valuesin  program

```python
df["Program"] = df["Program"].astype("string").str.strip()
df["Program"] = df["Program"].replace("---Select---", "Unknown")
df["Program"] = df["Program"].fillna("Unknown")
```

## Replace Invalid Values in year of pass 

```python
df["Year of Pass"] = df["Year of Pass"].fillna("Unknown")

df["Year of Pass"] = df["Year of Pass"].replace(
    ["", "--Select--", "nan"], "Unknown"
)
```

## Clean notes and descriptions

```python
df["Won Description"] = df["Won Description"].fillna("No Information")

df["Lost Description"] = df["Lost Description"].fillna("No Information")

df["Call Feedback"] = df["Call Feedback"].fillna("No Information")

df["Next Follow-up Notes"] = df["Next Follow-up Notes"].fillna("No Information")

df["Update few words about this lead"] = df[
    "Update few words about this lead"
].fillna("No Information")
```

## Clean marketing columns

```python
df["Campaign Name"] = df["Campaign Name"].fillna("No Campaign")

df["Campaign Term"] = df["Campaign Term"].fillna("Not Provided")

df["Campaign Content"] = df["Campaign Content"].fillna("Not Provided")

df["UTM"] = df["UTM"].fillna("Not Provided")

df["ad set name"] = df["ad set name"].fillna("No Ad Set")

df["ad name"] = df["ad name"].fillna("No Ad")
```

## Identifier / system fields

```python
df["Lead Automation Email"] = df["Lead Automation Email"].fillna("Not Provided")

df["Lead ID"] = df["Lead ID"].fillna(
    pd.Series(
        [f"LEAD_{i:05d}" for i in range(1, len(df) + 1)],
        index=df.index
    )
)
```

## Clean numeric columns

```python
df["Total_Outgoing_Calls"] = pd.to_numeric(
    df["Total_Outgoing_Calls"], errors="coerce"
).fillna(0)

df["Total_Incoming_Calls"] = pd.to_numeric(
    df["Total_Incoming_Calls"], errors="coerce"
).fillna(0)

df["Call Duration"] = pd.to_numeric(
    df["Call Duration"], errors="coerce"
).fillna(0)

```

## For revenue

```python
df["Won Revenue"] = pd.to_numeric(
    df["Won Revenue"], errors="coerce"
).fillna(0)

# If there are negative revenues:
df.loc[df["Won Revenue"] < 0, "Won Revenue"] = 0
```

## Clean Year of Pass

```python
df["Year of Pass"] = df["Year of Pass"].replace(
    "--Select--", "Unknown"
)
df["Year of Pass"] = df["Year of Pass"].fillna("Unknown")
```

## Clean date/time columns

```python
date_columns = [
    "Won Date",
    "Lead Date",
    "Expected Closing Date",
    "Lead Assigned At",
    "First Touch At",
    "Last Call Time",
    "Next Follow-up On",
    "Droped Date",
    "Created Time",
    "Updated Time",
    "Viewed Time"
]

# Convert columns to datetime
for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        errors="coerce"
    )
```

## Replace missing dates

```python
date_replacements = {
    "Won Date": "Not Won",
    "Lead Date": "Date Not Available",
    "Expected Closing Date": "Not Scheduled",
    "Lead Assigned At": "Not Assigned",
    "First Touch At": "No First Touch",
    "Last Call Time": "No Call Recorded",
    "Next Follow-up On": "No Follow-up",
    "Droped Date": "Not Dropped",
    "Created Time": "Not Available",
    "Updated Time": "Not Available",
    "Viewed Time": "Not Viewed"
}

for col, replacement in date_replacements.items():
    df[col] = df[col].dt.strftime("%Y-%m-%d")
    df[col] = df[col].fillna(replacement)
```

## Clean email

```python
df["Email Address"] = df["Email Address"].fillna("Not Provided")

df["Email Address"] = (
    df["Email Address"]
    .astype(str)
    .str.strip()
    .str.lower()
)
```

## Clean Call Missed by User

```python
df["Call Missed by User"] = df["Call Missed by User"].fillna("No")
```

# Final Cleaning Summary

### What was done

- Missing categorical values  → `Unknown`
- Missing email → `Not Provided`
- Missing campaign values  → `No Campaign`or  `Not Provided` 
- Missing ad set values  → `No Ad Set`
- Missing ad name values → `No Ad`
- Missing descriptions and feedback  → `No Information`
- Missing call metrics → `0`
- Missing won revenue → `0`
- Negative revenue values → `0`
- --Select-- replaced with  → `Unknown`
- -Other- standardized to Other.
- Date/time columns converted to datetime format.
- Placeholder dates used for missing date/time values.
- Duplicate rows checked but not automatically deleted.
- No rows deleted because of missing values.
- Final missing-value and dataset-size checks performed.

### Final workflow

**CRM Leads.csv → Inspect → Clean Missing Values → Convert Data Types → Validate → Export → Power BI**
